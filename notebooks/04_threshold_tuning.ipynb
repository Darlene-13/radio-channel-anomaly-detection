```
import numpy as np
from sklearn.metrics import f1_score
from pathlib import Path
```

```
data_dir = Path.cwd().parent /"data"/"processed"
y = np.load(data_dir/"y_train.npy")
oof_lr = np.load(data_dir/"oof_lr.npy")
oof_rf = np.load(data_dir/"oof_rf.npy")
oof_lgb = np.load(data_dir/"oof_lgb.npy")
```

```

def best_threshold(oof, y):
    best_f1, best_t = 0, 0.5
    for t in np.arange(0.05, 0.95, 0.01):
        f1 = f1_score(y, (oof > t).astype(int))
        if f1 > best_f1:
            best_f1, best_t = f1, t
    return best_f1, best_t


results = {}
for name, oof in [("Logistic Regression", oof_lr), ("Random Forest", oof_rf), ("LightGBM", oof_lgb)]:
    f1, t = best_threshold(oof, y)
    results[name] = (f1, t)
    print(f"{name:<20} best F1 = {f1:.4f}  at threshold = {t:.2f}")

winner = max(results, key=lambda k: results[k][0])
best_f1, best_t = results[winner]
print(f"\nWinner: {winner}  (F1 = {best_f1:.4f}, threshold = {best_t:.2f})")
```
